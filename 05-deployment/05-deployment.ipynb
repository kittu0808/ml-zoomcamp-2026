## Q3. Conversion Probability — Loaded Model

### Question
Load the official frozen machine learning model from `pipeline.bin` and calculate the conversion probability for the following lead.

### Lead Details

| Feature | Value |
|---|---|
| `lead_source` | `paid_ads` |
| `industry` | `technology` |
| `employment_status` | `employed` |
| `location` | `north_america` |
| `number_of_courses_viewed` | 2 |
| `annual_income` | 79276.0 |
| `interaction_count` | 4 |
| `lead_score` | 0.41 |

### Objective
Use the loaded model's `predict_proba()` method to calculate the probability that this lead will convert.

**Expected output format:** Conversion probability rounded to three decimal places.

### Approach
1. Load `pipeline.bin` using Python's `pickle` module.
2. Represent the lead as a dictionary using the specified feature names.
3. Call `predict_proba()` to obtain the conversion probability.
4. Round the probability to three decimal places and record the result.

````python
import pickle

# Load the official frozen model
with open("pipeline.bin", "rb") as f:
    model = pickle.load(f)

# Lead specified in Q3
lead = {
    "lead_source": "paid_ads",
    "industry": "technology",
    "employment_status": "employed",
    "location": "north_america",
    "number_of_courses_viewed": 2,
    "annual_income": 79276.0,
    "interaction_count": 4,
    "lead_score": 0.41
}

# Calculate conversion probability
probability = model.predict_proba([lead])[0, 1]

print(f"Conversion probability: {probability:.3f}")
````

### Output

```text
Conversion probability: 0.533
```

### Final Answer

**The conversion probability is 0.533** (approximately 53.3%).

### Interpretation

For the lead specified in Q3, the frozen model estimates a 53.3% probability of conversion. This is the model's predicted probability, not a guarantee that the lead will convert.
