```python
# Identify outliers using the IQR method

outlier_results = []

for column in numerical_features:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    outlier_results.append({
        'Feature': column,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound,
        'Outlier Count': len(outliers),
        'Outlier Percentage': len(outliers) / len(df) * 100
    })

outlier_df = pd.DataFrame(outlier_results)

display(outlier_df.round(3))
```

**Objective:** Prepare the dataset for machine learning by handling duplicate observations, separating input features from the target variable, splitting the data, and preparing appropriate preprocessing steps.

#### 6.1 Duplicate Removal

During exploratory data analysis, 125 additional duplicate occurrences were identified, involving 201 rows across duplicate groups.

All rows belonging to these duplicate groups had `Revenue = False`.

To avoid giving identical observations additional weight during model training, exact duplicate rows will be removed. The original dataset will be preserved, and a cleaned copy will be created for subsequent steps.

```python
# Create a copy of the original dataset
df_clean = df.copy()

# Remove exact duplicate rows
df_clean = df_clean.drop_duplicates()

# Reset the index
df_clean = df_clean.reset_index(drop=True)

print("Original dataset shape:", df.shape)
print("Dataset shape after duplicate removal:", df_clean.shape)

print("\nRemaining duplicate rows:", df_clean.duplicated().sum())

print("\nTarget distribution after duplicate removal:")
print(df_clean['Revenue'].value_counts())

print("\nTarget percentage after duplicate removal:")
print(
    df_clean['Revenue']
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

#### 6.2 Define Features and Target Variable

The target variable for this classification project is `Revenue`.

- `Revenue = True`: The visitor completed a purchase.
- `Revenue = False`: The visitor did not complete a purchase.

All other eligible columns will be used as input features.

The `PageValues` feature will initially be excluded from the primary model because its calculation may depend on transaction-related information. This helps us investigate a more realistic prediction scenario and reduce the risk of target leakage.

```python
# Define the target variable
y = df_clean['Revenue'].astype(int)

# Define the input features
X = df_clean.drop(columns=['Revenue', 'PageValues'])

print("Features shape:", X.shape)
print("Target shape:", y.shape)

print("\nTarget distribution:")
print(y.value_counts())

print("\nFeature columns:")
print(X.columns.tolist())
```

#### 6.3 Train-Test Split

The cleaned dataset is divided into training and testing subsets using an 80:20 ratio.

- The training dataset is used to train machine learning models.
- The testing dataset is used to evaluate model performance on unseen observations.

Stratified sampling is applied using the target variable `Revenue` to maintain approximately the same proportion of purchase and non-purchase sessions in both subsets.

A random state of 42 is used to ensure reproducibility.

```python
from sklearn.model_selection import train_test_split

# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Display dataset shapes
print("Training features shape:", X_train.shape)
print("Testing features shape:", X_test.shape)

print("\nTraining target shape:", y_train.shape)
print("Testing target shape:", y_test.shape)

# Check target distribution in training and testing sets
print("\nTraining target distribution (%):")
print(y_train.value_counts(normalize=True).mul(100).round(2))

print("\nTesting target distribution (%):")
print(y_test.value_counts(normalize=True).mul(100).round(2))
```

#### 6.4 Feature Preprocessing

Numerical and categorical features require different preprocessing techniques.

- Numerical features are standardized using StandardScaler.
- Categorical features are converted into numerical representations using OneHotEncoder.
- ColumnTransformer is used to apply these transformations to the appropriate columns.

The preprocessing pipeline is fitted using training data only to prevent data leakage from the test set.

```python
# Define numerical features
numerical_features_model = [
    'Administrative',
    'Administrative_Duration',
    'Informational',
    'Informational_Duration',
    'ProductRelated',
    'ProductRelated_Duration',
    'BounceRates',
    'ExitRates',
    'SpecialDay'
]

# Define categorical features
categorical_features_model = [
    'Month',
    'OperatingSystems',
    'Browser',
    'Region',
    'TrafficType',
    'VisitorType',
    'Weekend'
]

print("Number of numerical features:", len(numerical_features_model))
print("Number of categorical features:", len(categorical_features_model))
print("Total features:", len(numerical_features_model) + len(categorical_features_model))
```

```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# Numerical preprocessing
numerical_transformer = Pipeline(steps=[
    ('scaler', StandardScaler())
])

# Categorical preprocessing
categorical_transformer = Pipeline(steps=[
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])

# Combine both preprocessing pipelines
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numerical_transformer, numerical_features_model),
        ('cat', categorical_transformer, categorical_features_model)
    ]
)

print("Preprocessing pipeline created successfully.")
```

```python
# Fit preprocessing on training data and transform it
X_train_processed = preprocessor.fit_transform(X_train)

# Transform testing data using the already fitted preprocessor
X_test_processed = preprocessor.transform(X_test)

print("Processed training shape:", X_train_processed.shape)
print("Processed testing shape:", X_test_processed.shape)
```

```python
# Get transformed feature names
feature_names = preprocessor.get_feature_names_out()

print("Total transformed features:", len(feature_names))

print("\nFirst 20 transformed feature names:")
print(feature_names[:20])
```

#### 6.5 Baseline Model — Logistic Regression

Logistic Regression is selected as the first classification algorithm because it is simple, efficient, and suitable for binary classification.

A complete Scikit-learn pipeline is created by combining the preprocessing steps with Logistic Regression.

The `class_weight='balanced'` parameter is used to account for the imbalance between purchase and non-purchase sessions. The model is trained using the training dataset only.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

# Create the Logistic Regression pipeline
logistic_model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', LogisticRegression(
        class_weight='balanced',
        max_iter=1000,
        random_state=42
    ))
])

# Train the model
logistic_model.fit(X_train, y_train)

print("Logistic Regression model trained successfully!")
```

```python
# Predict purchase outcomes
y_pred = logistic_model.predict(X_test)

# Predict purchase probabilities
y_pred_proba = logistic_model.predict_proba(X_test)[:, 1]

print("First 10 actual values:")
print(y_test.iloc[:10].values)

print("\nFirst 10 predicted values:")
print(y_pred[:10])

print("\nFirst 10 predicted purchase probabilities:")
print(y_pred_proba[:10].round(3))
```

#### 6.6 Evaluation of Logistic Regression

The trained Logistic Regression model is evaluated using the testing dataset.

Since the target variable is imbalanced, accuracy alone is not sufficient. Therefore, precision, recall, F1-score, ROC-AUC, and the confusion matrix are used to obtain a more complete understanding of model performance.

These metrics help evaluate the model's ability to identify purchase sessions while controlling false purchase predictions.

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report
)

# Calculate evaluation metrics
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
roc_auc = roc_auc_score(y_test, y_pred_proba)

print("Logistic Regression - Evaluation Results")
print("----------------------------------------")
print(f"Accuracy:  {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall:    {recall:.4f}")
print(f"F1-Score:  {f1:.4f}")
print(f"ROC-AUC:   {roc_auc:.4f}")

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

### 6.8 Model Comparison

Two classification models, Logistic Regression and Random Forest, were trained and evaluated using the same training and testing datasets.

The results show that Random Forest achieved higher accuracy (83.57%), precision (45.70%), and ROC-AUC (78.51%). However, Logistic Regression achieved higher recall (78.27%) and F1-score (41.13%).

Logistic Regression identifies a larger proportion of actual purchasers, making it potentially more useful when the business objective is to identify as many potential customers as possible.

Random Forest is more conservative in predicting purchases and produces fewer false positives.

Since the dataset is imbalanced, accuracy alone is not sufficient to determine the best model. Recall, precision, F1-score, and ROC-AUC are also important evaluation metrics.

Further improvement through classification threshold tuning will be explored before selecting the final model.

```python
from sklearn.model_selection import train_test_split

X_train_sub, X_val, y_train_sub, y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.20,
    random_state=42,
    stratify=y_train
)

print("Training subset shape:", X_train_sub.shape)
print("Validation set shape:", X_val.shape)

print("\nTraining subset target distribution:")
print(y_train_sub.value_counts(normalize=True).round(4))

print("\nValidation target distribution:")
print(y_val.value_counts(normalize=True).round(4))
```

```python
rf_validation_model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', RandomForestClassifier(
        n_estimators=300,
        max_depth=None,
        min_samples_leaf=2,
        class_weight='balanced',
        random_state=42,
        n_jobs=-1
    ))
])

rf_validation_model.fit(X_train_sub, y_train_sub)

print("Random Forest validation model trained successfully!")
```

```python
val_probabilities = rf_validation_model.predict_proba(X_val)[:, 1]

print("First 10 actual values:")
print(y_val.values[:10])

print("\nFirst 10 predicted purchase probabilities:")
print(val_probabilities[:10].round(3))
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

thresholds = [0.20, 0.30, 0.40, 0.50, 0.60]

threshold_results = []

for threshold in thresholds:

    val_predictions = (
        val_probabilities >= threshold
    ).astype(int)

    threshold_results.append({
        'Threshold': threshold,
        'Accuracy': accuracy_score(y_val, val_predictions),
        'Precision': precision_score(
            y_val, val_predictions, zero_division=0
        ),
        'Recall': recall_score(
            y_val, val_predictions, zero_division=0
        ),
        'F1-Score': f1_score(
            y_val, val_predictions, zero_division=0
        )
    })

threshold_df = pd.DataFrame(threshold_results)

display(threshold_df.round(4))
```

```python
selected_threshold = 0.30

rf_tuned_y_pred = (
    rf_y_pred_proba >= selected_threshold
).astype(int)

print("Selected Threshold:", selected_threshold)

print("\nTuned Random Forest Metrics:")
print("Accuracy:", round(
    accuracy_score(y_test, rf_tuned_y_pred), 4
))
print("Precision:", round(
    precision_score(y_test, rf_tuned_y_pred), 4
))
print("Recall:", round(
    recall_score(y_test, rf_tuned_y_pred), 4
))
print("F1-Score:", round(
    f1_score(y_test, rf_tuned_y_pred), 4
))
```

### 6.9 Decision Threshold Tuning

The Random Forest model initially used the default classification threshold of 0.50. Since the dataset is imbalanced and identifying potential purchasers is an important objective, different classification thresholds were evaluated using a validation dataset.

The thresholds 0.20, 0.30, 0.40, 0.50, and 0.60 were compared.

The threshold of 0.30 achieved the highest validation F1-score of 0.4370, providing the best balance between precision and recall among the tested thresholds.

### 6.15 Limitations and Future Improvements

Although the developed model provides useful insights, several limitations should be considered.

**Limitations:**

1. The dataset contains an imbalanced target variable, with significantly more non-purchasing visitors than purchasing visitors.
2. The model's precision is relatively low, meaning some visitors predicted to purchase will not actually make a purchase.
3. The dataset represents a particular collection of online shopping sessions and may not generalize to every e-commerce platform.
4. The selected classification threshold was evaluated using a limited set of candidate values.
5. Random Forest feature importance indicates predictive relationships rather than causal relationships.
6. Model performance may change when applied to new customer populations or future website traffic.

**Future Improvements:**

1. Experiment with additional algorithms such as Gradient Boosting and XGBoost.
