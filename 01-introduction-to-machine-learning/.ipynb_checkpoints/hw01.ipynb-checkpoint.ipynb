## Question 6 — Median Horsepower

**Question:** Does the median value of `horsepower` change after filling the missing values with the median?

 The median is a useful measure of central tendency and is less affected by
 extreme values. We first calculate the original median, then fill the missing horsepower values with that median and calculate the median again.

```python
median_hp = df['horsepower'].median()
median_hp
```

```python
df['horsepower'].fillna(median_hp).median()
```

## Question 7 — Linear Algebra and Weight Calculation

**Question:** What is the sum of the calculated weights?

For this calculation, we use a subset of Asian cars and two features:
`vehicle_weight` and `model_year`.

The feature matrix is used to calculate the normal-equation solution:

$$
w = (X^T X)^{-1}X^T y
$$

This gives the model weights that best fit the provided target values `y`.

```python
asia = df[df['origin'] == 'Asia']

X = asia[['vehicle_weight', 'model_year']].head(7).values

XTX = X.T @ X

XTX_inv = np.linalg.inv(XTX)

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

w = XTX_inv @ X.T @ y

w.sum()
```
